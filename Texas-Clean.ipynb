```python
import pandas as pd
tx = pd.read_parquet('/content/drive/MyDrive/DSSI/NER Data/tx_transactors.parquet')
```

```python
def merge_name(row):

  if str(row['full_name']) != '<NA>':
    # print('name_preferred', type(row['name_preferred']))
    return row['full_name'].lower()
  elif str(row['name_preferred']) != '<NA>':
    # print('full_name', type(row['full_name']))
    return row['name_preferred'].lower()
  else:
    # print("else")
    name= str(row['first_name']) + ' ' + str(row['last_name'])
    name = name.replace('<NA>','').replace('  ',' ').strip().lower()
    return name
```

```python
tx['clean_name'] = tx.apply(merge_name, axis=1)
```

```python
pd_na = str(tx.iloc[0]['transactor_type'])
```

```python
pd_na
```

```python
import numpy as np
def replace_na(t_type, na=pd_na):
  if str(t_type) == pd_na:
    return np.nan
  else:
    return t_type
```

```python
tx['transactor_type'] = tx['transactor_type'].apply(replace_na)
```

```python
non_annot_tx = tx[tx['transactor_type'].isna()].copy()
```

```python
annot_tx = tx[tx['transactor_type'].notna()].copy()
```

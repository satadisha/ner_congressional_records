<h1>CONCAT ALL THE ANNOTATED DATA</h1>
<a href='https://pandas.pydata.org/docs/user_guide/merging.html'>REFERENCE</a>

```python
annotated_data = pa_transactor_annotated[['full_name', 'transactor_type']].copy()
annotated_data
```

```python
non_annotated_data = pa_transactors[pa_transactors['transactor_type'].isna()][['full_name','transactor_type']].copy()
non_annotated_data
```

```python
pa_annotated_2 = pa_transactors[pa_transactors['transactor_type'].notna()][['full_name','transactor_type']].copy()
pa_annotated_2
```

```python
az_non_annotated = az_transactors[az_transactors['transactor_type'].isna()][['full_name','transactor_type']].copy()
az_non_annotated
```

```python
az_annotated_2 = az_transactors[az_transactors['transactor_type'].notna()][['full_name','transactor_type']].copy()
az_annotated_2
```

```python
annotated_data = pd.concat([annotated_data, pa_annotated_2, az_annotated_2])
annotated_data.drop_duplicates(subset=['full_name'], inplace=True)
annotated_data.groupby(by='full_name').size().sort_values()
```

```python
non_annotated_data = pd.concat([non_annotated_data, az_non_annotated])
non_annotated_data.drop_duplicates(subset=['full_name'], inplace=True)
non_annotated_data.groupby(by='full_name').size().sort_values()
```

```python
annotated_data.to_csv('annotated_data.csv')
non_annotated_data.to_csv('non_annotated_data.csv')
```

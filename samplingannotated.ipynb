### Over Sampling Annotated (PA)

```python
import pandas as pd
from imblearn.over_sampling import RandomOverSampler

# load annotated dataset
df = pd.read_csv("/content/gdrive/MyDrive/DSSI Data/pa_transactor_annotated_clean.csv")

# clean string columns
for col in ['transactor_type', 'reported_state', 'party', 'full_name']:
    df[col] = df[col].astype(str).str.strip().str.title()

# replace filler or unknown values with NA
filler_values = ['Na', 'N/A', 'None', 'Unknown', 'Nan']
df.replace(filler_values, pd.NA, inplace=True)

# drop rows with missing key fields
df.dropna(subset=['full_name', 'transactor_type', 'reported_state', 'party'], inplace=True)

# drop duplicate donors based on id or name
df = df.drop_duplicates(subset=['id', 'full_name'])

# encode transactor_type as numeric labels
df['label'] = df['transactor_type'].factorize()[0]

# features and target for over-sampling
X = df[['transactor_type']]
y = df['label']

# apply random over-sampling to balance the classes
ros = RandomOverSampler(random_state=42)
X_resampled, y_resampled = ros.fit_resample(X, y)

# merge resampled data with full original data
df_resampled = pd.merge(
    pd.concat([X_resampled, y_resampled.rename("label")], axis=1),
    df[['id', 'full_name', 'transactor_type', 'reported_state', 'party', 'label']],
    on=['transactor_type', 'label'],
    how='left'
).drop_duplicates().reset_index(drop=True)

# print class distribution and sample data
print("Balanced transactor_type counts (after over-sampling):")
print(df_resampled['transactor_type'].value_counts())

print("Sample of cleaned and balanced data:")
print(df_resampled[['full_name', 'transactor_type', 'reported_state', 'party']].head())
```

### Over Sampling Annotated (AZ)

```python
import pandas as pd
from imblearn.over_sampling import RandomOverSampler

# load dataset
df = pd.read_csv("/content/gdrive/MyDrive/DSSI Data/az_transactors_clean.csv")

# clean transactor_type
df['transactor_type'] = df['transactor_type'].astype(str).str.strip().str.title()

# drop rows missing transactor_type or full_name
df = df.dropna(subset=['transactor_type', 'full_name'])

# remove duplicates based on id or full_name
df = df.drop_duplicates(subset=['id', 'full_name'])

# encode transactor_type
df['label'] = df['transactor_type'].factorize()[0]

# features and target
X = df[['transactor_type']]
y = df['label']

# apply random over-sampling
ros = RandomOverSampler(random_state=42)
X_resampled, y_resampled = ros.fit_resample(X, y)

# merge back full data
df_resampled = pd.merge(
    pd.concat([X_resampled, y_resampled.rename('label')], axis=1),
    df[['id', 'transactor_type', 'last_name', 'first_name', 'middle_name', 'name_suffix', 'full_name', 'label']],
    on=['transactor_type', 'label'],
    how='left'
).drop_duplicates().reset_index(drop=True)

# print balanced class counts
print("After over-sampling class counts:")
print(df_resampled['transactor_type'].value_counts())

# print sample output
print(df_resampled[['full_name', 'transactor_type']].head())
```

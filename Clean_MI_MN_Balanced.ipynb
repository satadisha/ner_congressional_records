```python
import pandas as pd
mi_mn_df = pd.read_csv('/content/drive/MyDrive/DSSI/NER Data/mi-mn-balanced.csv')
```

```python
def remove_duplicate_words(name):
    if not isinstance(name, str):
        return name

    words = name.strip().split()
    seen = set()
    result = []

    for word in words:
        if word.lower() not in seen:  # Case-insensitive match
            seen.add(word.lower())
            result.append(word)

    return ' '.join(result)

# Apply to your DataFrame
mi_mn_df['full_name'] = mi_mn_df['full_name'].apply(remove_duplicate_words)
```

```python
mi_mn_df.to_csv('clean-mi-mn-balanced.csv', index=False)
```
